## **Smart Electric Vehicle Battery Performance Analysis**

```python
import pandas as pd

data = {
    'Time (h)': [0, 1, 2, 3, 4, 5],
    'Temperature (°C)': [32, 35, 38, 40, 42, 45],
    'Voltage (V)': [400, 395, 390, 385, 380, 375],
    'Current (A)': [20, 22, 25, 28, 30, 32],
    'Charge (%)': [95, 88, 80, 70, 60, 48]
}

df = pd.DataFrame(data)

display(df)
```

Task 1 — Analysis Using Python Basics
Store the given sensor readings using appropriate Python data structures, then perform the following tasks.
1. Display the sensor readings
Use a for loop to display all battery readings.
2. Identify high temperature
Apply the following condition:
  If  T > 40 °C  
Display: “High Temperature Alert”
Otherwise, display: “Temperature Normal”
3. Identify low battery
Apply the following condition:
  If  C < 50 %  
Display: “Low Battery Alert”
Otherwise, display: “Battery Charge Normal”
4. Create a function
Create a function battery_status(charge) that returns a status based on:
           High,     C > 70
Status =   Medium,   50 ≤ C ≤ 70
           Low,      C < 50


```python
sensor_readings = []
for i in range(len(data['Time (h)'])):
    sensor_readings.append({
        'time': data['Time (h)'][i],
        'temperature': data['Temperature (°C)'][i],
        'voltage': data['Voltage (V)'][i],
        'current': data['Current (A)'][i],
        'battery_charge': data['Charge (%)'][i]
    })

print("Sensor readings prepared for basic Python analysis:")
for reading in sensor_readings:
    print(reading)
```

### 1. Display all battery readings

```python
print("Battery Readings:")
for reading in sensor_readings:
    print(f"  {reading['battery_charge']}% Charge")
```

### 2. Identify high temperature

```python
print("\nTemperature Status:")
for i, reading in enumerate(sensor_readings):
    if reading['temperature'] > 40:
        print(f"  Sensor {i+1}: High Temperature Alert (T = {reading['temperature']} °C)")
    else:
        print(f"  Sensor {i+1}: Temperature Normal (T = {reading['temperature']} °C)")
```

### 3. Identify low battery

```python
print("\nBattery Charge Status:")
for i, reading in enumerate(sensor_readings):
    if reading['battery_charge'] < 50:
        print(f"  Sensor {i+1}: Low Battery Alert (C = {reading['battery_charge']} %)")
    else:
        print(f"  Sensor {i+1}: Battery Charge Normal (C = {reading['battery_charge']} %)")
```

### 4. Create a function `battery_status(charge)`

```python
def battery_status(charge):
    if charge > 70:
        return "High"
    elif 50 <= charge <= 70:
        return "Medium"
    else:
        return "Low"

print("\nBattery Status Function Test:")
for i, reading in enumerate(sensor_readings):
    status = battery_status(reading['battery_charge'])
    print(f"  Sensor {i+1}: Charge {reading['battery_charge']}%, Status: {status}")
```

The company uses temperature and current to calculate a simplified Battery Stress Index.
Let the transformation matrix be:
      ┌ 2   1 ┐
A  =  │       │

Consider the sensor input:
      ┌ T ┐     ┌ 40 ┐
X  =  │   │  =  │    │

The output is calculated as:
  Y  =  A X  
Therefore:
      ┌ 2   1 ┐   ┌ 40 ┐
Y  =  │       │ × │    │
      └ 1   3 ┘   └ 25 ┘
Students must calculate the resulting vector Y.
Additional Linear Algebra Tasks
For the matrix A above, find:
1. Transpose
  Aᵀ  
2. Determinant
  det(A)  
3. Inverse
A⁻¹, if the inverse exists.
Solving a System of Equations
The engineers need to determine two unknown battery parameters x and y. The equations are:
  2x + y = 100  
  x + 3y = 120  
Write the equations in matrix form:
  A X = B  
where:
      ┌ 2   1 ┐        ┌ x ┐        ┌ 100 ┐
A  =  │       │   X =  │   │   B  =  │     │
      └ 1   3 ┘        └ y ┘        └ 120 ┘
Students must solve for x and y using NumPy.


```python
import numpy as np

A = np.array([
    [2, 1],
    [1, 3]
])

X = np.array([
    [40],
    [25]
])

print("Matrix A:")
print(A)
print("\nVector X:")
print(X)
```

```python
Y = np.dot(A, X)

print("Resulting Vector Y:")
print(Y)
```

### 1. Transpose of A

```python
A_transpose = A.T

print("Transpose of Matrix A (Aᵀ):")
print(A_transpose)
```

### 2. Determinant of A

```python
A_determinant = np.linalg.det(A)

print("Determinant of Matrix A (det(A)):")
print(A_determinant)
```

### 3. Inverse of A

```python
A_inverse = np.linalg.inv(A)

print("Inverse of Matrix A (A⁻¹):")
print(A_inverse)
```

### Solving a System of Equations

```python
A_eq = np.array([
    [2, 1],
    [1, 3]
])

B_eq = np.array([
    [100],
    [120]
])

solution = np.linalg.solve(A_eq, B_eq)

print("Solution for the system of equations (x, y):")
print(solution)
```

Task 4 — Analysis Using Calculus
The battery charge is modeled as a function of time:
  C(t)  =  100 − 2t − ½t²  
where C(t) is the battery charge percentage and t is the time in hours.
1. Find the rate of change of battery charge
The rate at which the battery charge changes is given by dC/dt. Differentiate C(t) and determine dC/dt.
2. Calculate the discharge rate
Find the discharge rate at t = 2 and t = 5, that is, calculate:
  dC/dt  at  t = 2      and      dC/dt  at  t = 5  
3. Interpret the result
Based on the values of dC/dt, determine whether the battery is losing charge:
At a constant rate
At an increasing rate
At a decreasing rate


### 1. Find the rate of change of battery charge

```python
from sympy import symbols, diff

t = symbols('t')
C = 100 - 2*t - (1/2)*t**2
dC_dt = diff(C, t)

print(f"The rate of change of battery charge, dC/dt, is: {dC_dt}")
```

### 2. Calculate the discharge rate

```python
discharge_rate_t2 = dC_dt.subs(t, 2)
discharge_rate_t5 = dC_dt.subs(t, 5)

print(f"Discharge rate at t = 2 hours: {discharge_rate_t2} %/hour")
print(f"Discharge rate at t = 5 hours: {discharge_rate_t5} %/hour")
```
