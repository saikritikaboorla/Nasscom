## 2. Handle Missing Values

```python
# -----------------------------------------------------------
# 🔹 2A. DEMO — find and fill missing values
# -----------------------------------------------------------
print('Missing values per column:')
print(df.isna().sum().sort_values(ascending=False).head())

df['age'] = df['age'].fillna(df['age'].median())
print('\nMissing ages after fillna:', df['age'].isna().sum())
```

#### 🧪 LAB EXERCISE 2 — Clean missing values

Using `lab`:
1. Print how many values are missing in the **`deck`** column.
2. Drop the **`deck`** column entirely (too many missing values).
3. Fill missing **`embark_town`** values with the mode.
4. Confirm there are no missing values left in `embark_town`.

```python
# 1. Missing values in 'deck'
print('Missing in deck:', lab['deck'].isna().sum())

# 2. Drop 'deck' column
lab = lab.drop(columns=['deck'])

# 3. Fill missing 'embark_town' with the mode (hint: lab['embark_town'].mode()[0])
mode_embark_town = lab['embark_town'].mode()[0]
lab['embark_town'] = lab['embark_town'].fillna(mode_embark_town)

# 4. Confirm
print('Missing in embark_town after fillna:', lab['embark_town'].isna().sum())
```

## 3. Filtering & Sorting

```python
# -----------------------------------------------------------
# 🔹 3A. DEMO — filter rows and sort values
# -----------------------------------------------------------
first_class = df[df['class'] == 'First']
print('First class passengers:', len(first_class))

top_fares = df.sort_values('fare', ascending=False).head(5)
top_fares[['fare', 'class', 'sex']]
```

#### 🧪 LAB EXERCISE 3 — Filter and sort

Using `lab`:

2. Sort `survivors_f` by `age` ascending and print the youngest 5.

## 4. Group By (split → apply → combine)

```python
# -----------------------------------------------------------
# 🔹 4A. DEMO — groupby with one and two keys
# -----------------------------------------------------------
print('Survival rate by class:')
print(df.groupby('class', observed=True)['survived'].mean())

print('\nSurvival rate by class and sex:')
print(df.groupby(['class', 'sex'], observed=True)['survived'].mean())
```

#### 🧪 LAB EXERCISE 4 — Group and aggregate

Using `lab`:
1. Group by `pclass` and print the mean `fare` per class.
2. Group by `sex` and print both the mean AND max `age` per group (hint: `.agg(['mean', 'max'])`).
3. Group by `embark_town` and print the total number of passengers per town (hint: `.size()` or `.count()`).

```python
# 1. Mean fare by pclass
print('Mean fare by pclass:')
print(lab.groupby('pclass')['fare'].mean())

# 2. Age mean & max by sex
print('Age stats by sex:')
print(lab.groupby('sex')['age'].agg(['mean', 'max']))

# 3. Passenger count by embark_town
print('Passenger count by embark_town:')
print(lab.groupby('embark_town').size())
```

## 5. Apply & Lambda (custom transformations)

```python
# -----------------------------------------------------------
# 🔹 5A. DEMO — apply a custom function to a column
# -----------------------------------------------------------
df['fare_category'] = df['fare'].apply(
    lambda f: 'high' if f > 50 else ('medium' if f > 15 else 'low')
)
print(df['fare_category'].value_counts())
```

#### 🧪 LAB EXERCISE 5 — Custom columns with apply

Using `lab`:
1. Create a new column `age_group` that labels passengers as `'child'` (age < 18), `'adult'` (18-59), or `'senior'` (60+) using `.apply()`.
2. Print the `value_counts()` of `age_group`.
3. Create a column `family_size` = `sibsp` + `parch` + 1, then print the average survival rate grouped by `family_size`.

```python
# 1. age_group via apply
def get_age_group(age):
    if age < 18:
        return 'child'
    elif 18 <= age < 60:
        return 'adult'
    else:
        return 'senior'
lab['age_group'] = lab['age'].apply(get_age_group)

# 2. value_counts of age_group
print('Age group counts:')
print(lab['age_group'].value_counts())

# 3. family_size and survival rate by family_size
lab['family_size'] = lab['sibsp'] + lab['parch'] + 1
print('Survival rate by family size:')
print(lab.groupby('family_size')['survived'].mean())
```

## 6. Pivot Tables & Merge/Join

```python
# -----------------------------------------------------------
# 🔹 6A. DEMO — pivot_table and merge
# -----------------------------------------------------------
pivot = df.pivot_table(
    values='survived', index='class', columns='sex',
    aggfunc='mean', observed=True
)
print('Survival rate pivot (class x sex):')
print(pivot)

left = pd.DataFrame({'id': [1, 2, 3], 'name': ['A', 'B', 'C']})
right = pd.DataFrame({'id': [1, 2, 4], 'score': [90, 85, 70]})
merged = left.merge(right, on='id', how='left')
print('\nLeft join example:')
print(merged)
```
