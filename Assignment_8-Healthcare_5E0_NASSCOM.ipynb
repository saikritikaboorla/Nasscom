## **Healthcare Case Study**

```python
import pandas as pd

data = {
    'Time (h)': [0, 1, 2, 3, 4, 5],
    'Heart Rate': [76, 82, 90, 98, 105, 112],
    'Temperature': [36.7, 37.0, 37.4, 38.0, 38.5, 39.0],
    'SpO₂': [98, 97, 96, 94, 92, 89],
    'Systolic BP': [122, 125, 130, 135, 140, 148],
    'Health Score': [92, 88, 82, 74, 65, 52]
}

df = pd.DataFrame(data)
display(df.head())
```

Task 1 — Analysis Using Python Basics
Store the patient readings using appropriate Python data structures, then perform the following tasks.
1. Display the Patient Readings
Use a for loop to display all patient readings.
2. Identify High Temperature
If Temperature > 38°C, display “High Temperature Alert”. Otherwise, display “Temperature Normal”.
3. Identify Low Oxygen Level

### Task 1.1: Display Patient Readings

```python
for index, row in df.iterrows():
    print(f"Time: {row['Time (h)']:.0f}h, Heart Rate: {row['Heart Rate']:.0f} BPM, Temperature: {row['Temperature']:.1f}°C, SpO₂: {row['SpO₂']:.0f}%, Systolic BP: {row['Systolic BP']:.0f} mmHg, Health Score: {row['Health Score']:.0f}%")
```

### Task 1.2: Identify High Temperature

```python
for temp in df['Temperature']:
    if temp > 38:
        print(f"High Temperature Alert: {temp}°C")
    else:
        print(f"Temperature Normal: {temp}°C")
```

### Task 1.3: Identify Low Oxygen Level

```python
for spo2 in df['SpO₂']:
    if spo2 < 90:
        print(f"Low Oxygen Alert: {spo2}%")
    else:
        print(f"Oxygen Level Normal: {spo2}%")
```

### Task 1.4: Create a Health Status Function

```python
def health_status(score):
    if score > 80:
        return "Healthy"
    elif 50 <= score <= 80:
        return "Monitor"
    else:
        return "Critical"

print(f"Health status for score 92: {health_status(92)}")
print(f"Health status for score 74: {health_status(74)}")
print(f"Health status for score 52: {health_status(52)}")
print(f"Health status for score 45: {health_status(45)}")
```

Task 3 — Analysis Using Linear Algebra
The hospital uses a simplified Patient Risk Index based on heart rate and temperature.
Let A = [[2, 1], [1, 3]] and X = [[98], [38]]. Calculate Y = AX and the resulting vector Y.
Additional Linear Algebra Tasks: For matrix A, find (1) transpose Aᵀ, (2) determinant det(A), and (3) inverse A⁻¹, if it exists.
Solving a System of Equations
Solve: 2x + y = 100 and x + 3y = 120. Write the equations in matrix form AX = B and use NumPy to solve for x and y.


### Task 3.1: Calculate Patient Risk Index Y = AX

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
X = np.array([[98], [38]])

Y = np.dot(A, X)

print("Matrix A:\n", A)
print("Matrix X:\n", X)
print("Resulting vector Y (AX):\n", Y)
```

### Task 3.2: Additional Linear Algebra Tasks for Matrix A

```python
A_transpose = A.T
det_A = np.linalg.det(A)

print("Transpose of A (Aᵀ):\n", A_transpose)
print("Determinant of A (det(A)):", det_A)

if det_A != 0:
    A_inverse = np.linalg.inv(A)
    print("Inverse of A (A⁻¹):\n", A_inverse)
else:
    print("Inverse of A does not exist as determinant is zero.")
```

### Task 3.3: Solving a System of Equations (AX = B)

```python
A_eq = np.array([[2, 1], [1, 3]])
B_eq = np.array([[100], [120]])

x_y_solution = np.linalg.solve(A_eq, B_eq)

print("Matrix A for equations:\n", A_eq)
print("Matrix B for equations:\n", B_eq)
print("Solution for x and y (x, y):\n", x_y_solution)
```

Task 4 — Analysis Using Calculus
The patient's health score is modeled as H(t) = 100 − 4t − ½t², where H(t) is the health score and t is time in hours.
1. Differentiate H(t) and determine dH/dt.
2. Calculate dH/dt at t = 2 and t = 5.
3. Interpret whether the health score is decreasing at a constant, increasing, or decreasing rate.


### Task 4.1: Differentiate H(t) and Determine dH/dt

```python
import sympy
from sympy import symbols, diff

t = symbols('t')
H_t = 100 - 4*t - (1/2)*t**2
dH_dt = diff(H_t, t)

print(f"H(t) = {H_t}")
print(f"dH/dt = {dH_dt}")
```

### Task 4.2: Calculate dH/dt at t = 2 and t = 5

```python
dH_dt_at_2 = dH_dt.subs(t, 2)
dH_dt_at_5 = dH_dt.subs(t, 5)

print(f"dH/dt at t = 2: {dH_dt_at_2}")
print(f"dH/dt at t = 5: {dH_dt_at_5}")
```

### Task 4.3: Interpret the Rate of Health Score Change

```python
if dH_dt.is_constant():
    rate_type = "constant"
elif diff(dH_dt, t) > 0:
    rate_type = "increasing"
elif diff(dH_dt, t) < 0:
    rate_type = "decreasing"
else:
    rate_type = "varying"

print(f"The health score is decreasing at a {rate_type} rate, as dH/dt is a linear function of t, specifically -4 - t, which becomes more negative as t increases.")
```
