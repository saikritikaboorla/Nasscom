## **Agriculture Case Study**

```python
import pandas as pd

data = {
    'Day': [1, 2, 3, 4, 5, 6],
    'Soil Moisture': [72, 68, 62, 55, 48, 40],
    'Temperature': [27, 29, 31, 33, 35, 37],
    'Humidity': [78, 74, 70, 65, 60, 55],
    'Nitrogen': [48, 46, 43, 39, 35, 30],
    'Crop Health': [94, 90, 84, 76, 65, 52]
}

df = pd.DataFrame(data)
display(df.head())
```

Task 1 — Analysis Using Python Basics
Store the crop-monitoring readings using appropriate Python data structures, then perform the following tasks.
1. Display the Crop Readings
Use a for loop to display all readings.
2. Identify Low Soil Moisture
If Soil Moisture < 50%, display “Low Soil Moisture Alert”. Otherwise, display “Soil Moisture Normal”.
3. Identify High Temperature
If Temperature > 35°C, display “High Temperature Alert”. Otherwise, display “Temperature Normal”.
4. Create a Function
Create crop_status(score) that returns: Healthy (score > 80), Under Stress (50 ≤ score ≤ 80), Critical (score < 50).


```python
for index, row in df.iterrows():
    print(f"Day {int(row['Day'])}: Soil Moisture={row['Soil Moisture']}%, Temperature={row['Temperature']}°C, Humidity={row['Humidity']}%, Nitrogen={row['Nitrogen']} mg/kg, Crop Health={row['Crop Health']}%")
```

```python
for index, row in df.iterrows():
    if row['Soil Moisture'] < 50:
        print(f"Day {int(row['Day'])}: Low Soil Moisture Alert")
    else:
        print(f"Day {int(row['Day'])}: Soil Moisture Normal")
```

```python
for index, row in df.iterrows():
    if row['Temperature'] > 35:
        print(f"Day {int(row['Day'])}: High Temperature Alert")
    else:
        print(f"Day {int(row['Day'])}: Temperature Normal")
```

```python
def crop_status(score):
    if score > 80:
        return "Healthy"
    elif 50 <= score <= 80:
        return "Under Stress"
    else:
        return "Critical"

for index, row in df.iterrows():
    health_score = row['Crop Health']
    status = crop_status(health_score)
    print(f"Day {int(row['Day'])} Crop Health Score: {health_score}%, Status: {status}")
```

Task 3 — Analysis Using Linear Algebra
The agriculture system uses soil moisture and temperature to calculate a simplified Crop Stress Index.
Let A = [[2, 1], [1, 3]] and X = [[55], [33]]. Calculate Y = AX and the resulting vector Y.
Additional Linear Algebra Tasks: For matrix A, find (1) transpose Aᵀ, (2) determinant det(A), and (3) inverse A⁻¹, if it exists.
Solving a System of Equations
Solve: 2x + y = 100 and x + 3y = 120. Write the equations in matrix form AX = B and use NumPy to solve for x and y.


```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
X = np.array([[55], [33]])

Y = A @ X
print("Matrix A:")
print(A)
print("Matrix X:")
print(X)
print("Resulting vector Y = AX:")
print(Y)
```

```python
A = np.array([[2, 1], [1, 3]])

transpose_A = A.T
determinant_A = np.linalg.det(A)
inverse_A = np.linalg.inv(A)

print("Transpose of A (Aᵀ):")
print(transpose_A)
print("Determinant of A (det(A)):")
print(determinant_A)
print("Inverse of A (A⁻¹):")
print(inverse_A)
```

```python
A_system = np.array([[2, 1], [1, 3]])
B_system = np.array([100, 120])

solution = np.linalg.solve(A_system, B_system)

x = solution[0]
y = solution[1]

print("System of equations: 2x + y = 100, x + 3y = 120")
print("Matrix A for system:")
print(A_system)
print("Vector B for system:")
print(B_system)
print("Solution for x and y:")
print(f"x = {x:.2f}")
print(f"y = {y:.2f}")
```
